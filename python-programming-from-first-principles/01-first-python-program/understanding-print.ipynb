## Data, code, and output

**Programming** means designing and writing instructions, called **code**, that tell a computer what to do. These instructions work with **data**: facts or ideas represented as text, numbers, or other forms a computer can process.

In our announcement, the event title and capacity are data. The `print()` instructions display them as **output**. That output conveys **information**—meaning in context: `Planned capacity: 120` tells us that the event has 120 planned places.[^data-information] The title line shows the same connection:

```python
print("Community Storytelling Evening")
```

The title `Community Storytelling Evening` is the data. The complete `print()` instruction is the code that displays it, producing this output:

```text
Community Storytelling Evening
```

Python displays the title we supplied. Changing the text inside the quotation marks changes the output; it does not change what `print()` does.

## A note for the reader

The program also contains a line that produces no output:

```python
# Publish a community event announcement.
```

This is a **comment**, a note for someone reading the code. Outside quotation marks, `#` begins a comment that Python ignores for the rest of that line. Here, it explains why the program was written. That is why the program has six written lines but displays only five.

## Two values on one line

The final instruction differs from the others because it supplies two arguments:

```python
print("Planned capacity:", 120)
```

The comma separates the label `"Planned capacity:"` from the number `120`. With this call, `print()` displays both values on one line, putting a space between them:

```text
Planned capacity: 120
```

The label tells the reader what the number means. The comma belongs to the instruction and is not displayed.


[^data-information]: Definitions paraphrased from the NIST Computer Security Resource Center glossary, “Data” (https://csrc.nist.gov/glossary/term/data) and “Information” (https://csrc.nist.gov/glossary/term/information), drawing on its representation and contextual-meaning definitions.
