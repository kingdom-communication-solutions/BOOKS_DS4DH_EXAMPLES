## Data, code, and output

The announcement shows how instructions and information work together. The written instructions are **code**. They use **data**: facts or ideas represented in a form a computer can process, such as the event title written as text or the capacity written as a number.

```python
print("Community Storytelling Evening")
```

Here, `"Community Storytelling Evening"` is the data, and the complete instruction is the code that displays it. The displayed result is **output**:

```text
Community Storytelling Evening
```

## A note for the reader

One line in the program produces no output:

```python
# Publish a community event announcement.
```

This is a **comment**, a note explaining the program’s purpose to a human reader. Outside quotation marks, `#` starts a comment that Python ignores for the rest of the line. Our six-line program therefore displays only five lines.

## Two values on one line

The last instruction supplies two arguments rather than one:

```python
print("Planned capacity:", 120)
```

The comma separates the text label from the number. `print()` displays them on one line with a space between them:

```text
Planned capacity: 120
```

The label gives the number a context: `120` refers to planned places at the event. This meaning is **information**. The comma separates arguments in the code; it does not appear in the output.
