## Understanding print()

A title, city, date, greeting, and capacity are **data**: information represented for a purpose. Python instructions are **code**. What appears after Python follows an instruction is **output**. ## Explain

`print` is a Python **function**: a named operation that performs a task. Writing `print(...)` asks Python to perform that task, so the complete instruction is a **function call**. A value supplied between the parentheses is an **argument**.

The first line begins with `#`, outside quotation marks. It is a **comment**, a note for a human reader; Python does not execute it.
